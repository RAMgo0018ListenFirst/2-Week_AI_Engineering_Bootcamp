```python
###create a simple txt file

import os
os.makedirs("../data/text_file", exist_ok=True)
```

```python
sample_text={"../data/text_file/python_intro.txt":"""This the Content of the python intro file. It contains basic information about python programming language. Python is a high-level, interpreted programming language that emphasizes code readability and simplicity. It was created by Guido van Rossum and first released in 1991. Python supports multiple programming paradigms, including procedural, object-oriented, and functional programming. It has a large standard library that provides a wide range of modules and functions for various tasks, making it a popular choice for web development, data analysis, artificial intelligence, scientific computing, and more. Python's syntax is designed to be easy to read and write, which contributes to its popularity among both beginners and experienced programmers.""",


"../data/text_file/python_machine_learning.txt":"""This the Content of the python advanced file. It contains advanced information about python programming language. Python is a versatile programming language that offers a wide range of features and capabilities for advanced users. It supports advanced data structures such as lists, tuples, sets, and dictionaries, which allow for efficient data manipulation and storage. Python also provides powerful libraries and frameworks for various applications, including machine learning (e.g., TensorFlow, scikit-learn), web development (e.g., Django, Flask), and scientific computing (e.g., NumPy, SciPy). Additionally, Python has a strong community that contributes to its ecosystem with numerous third-party packages and tools. Advanced users can take advantage of Python's dynamic typing, decorators, generators, and context managers to write more efficient and elegant code.""",
}
for file_path, content in sample_text.items():
    with open(file_path, "w") as f:
        f.write(content)
print("Sample text files created successfully.")
```

```python
### TextLoader
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/text_file/python_intro.txt",encoding="utf-8")
document=loader.load()
print(document)

```

```python
###directory loader
from langchain_community.document_loaders import DirectoryLoader

###load all text files from the directory
dir_loader=DirectoryLoader("../data/text_file", 
glob="**/*.txt", #Pattern to match files
loader_cls=TextLoader,#loader class to use for loading files
loader_kwargs={"encoding":"utf-8"},
show_progress=False,)
documents=dir_loader.load()
documents
```
